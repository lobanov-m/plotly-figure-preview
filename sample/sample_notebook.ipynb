```python
import numpy as np
from PIL import Image
import plotly.graph_objects as go

rng = np.random.default_rng(0)
```

## uint8 RGB — shown as-is

Try: `rgb`, `rgb[..., 0]`, `rgb[:60, :80]`


```python
y, x = np.mgrid[0:240, 0:320]
rgb = np.dstack([
    (x / 320 * 255).astype(np.uint8),
    (y / 240 * 255).astype(np.uint8),
    ((np.sin(x / 18) * np.cos(y / 18) + 1) * 127).astype(np.uint8),
]).astype(np.uint8)
rgb.shape, rgb.dtype
```

## float32 in metres — normalized by its own min/max

Try: `depth`. The status bar reports the range it used.


```python
depth = (0.8 + np.hypot(x - 160, y - 120) / 40).astype(np.float32)
float(depth.min()), float(depth.max())
```

## A batch — preview one element with an expression

Try: `batch[2]`, which the Variables pane could never name on its own.


```python
batch = np.stack([np.roll(rgb, i * 40, axis=1) for i in range(4)])
batch.shape
```

## PIL and Plotly

Try: `pil`, `fig`


```python
pil = Image.fromarray(rgb).convert('P')  # palette image, expanded to RGBA for display
fig = go.Figure(data=[go.Scatter(y=np.sin(np.linspace(0, 12, 200)))], layout={'title': 'sine'})
type(pil).__name__, type(fig).__name__
```
